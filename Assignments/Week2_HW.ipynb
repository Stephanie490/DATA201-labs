```python
#8g
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(8, 8))

college.hist('Grad.Rate', color='grey', bins=12, ax=axes[0, 0])
axes[0,0].set_title('Graduation Rate')

college.hist('Room.Board', color='cyan', bins=12, ax=axes[0, 1])
axes[0,1].set_title('Room and Board Cost')

college.hist('Top25perc', color='blue', bins=12, ax=axes[1,0])
axes[1,0].set_title('New students were top 25% of HS Class')

college.hist('F.Undergrad', color='green', bins=12, ax=axes[1,1])
axes[1,1].set_title('# of Full-time Undergraduates')
```
